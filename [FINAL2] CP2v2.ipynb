```python
# =========================================================
# IMPORT LIBRARIES
# =========================================================
from pathlib import Path
import os

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.mixture import GaussianMixture

import statsmodels.formula.api as smf
from statsmodels.stats.multicomp import pairwise_tukeyhsd

from scipy.stats import f_oneway

from pygam import LinearGAM, s

# =========================================================
# SAVE ALL FIGURES
# =========================================================
FIG_DIR = Path("figures")
FIG_DIR.mkdir(exist_ok=True)

def save_fig(name):
    plt.savefig(FIG_DIR / f"{name}.png", dpi=300, bbox_inches="tight")

# =========================================================
# LOAD DATASET
# =========================================================
df = pd.read_csv("../Data/human_cognitive_performance.csv")

# =========================================================
# DATA UNDERSTANDING
# =========================================================
print("\n" + "="*60)
print("DATASET OVERVIEW")
print("="*60)

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDescriptive Statistics:")
print(df.describe())

#check categorical variables
print("\nGender Distribution:")
print(df['Gender'].value_counts())

print("\nExercise Frequency Distribution:")
print(df['Exercise_Frequency'].value_counts())

print("\nDiet Type Distribution:")
print(df['Diet_Type'].value_counts())

# =========================================================
# CHAPTER 3.4: DATA PREPROCESSING
# =========================================================
# 3.4.1 Remove Unused Variable
df.drop(columns=['User_ID', 'AI_Predicted_Score'], inplace=True)

# 3.4.2 Encode Categorical Variables
# gender
df['Gender'] = df['Gender'].map({
    'Male': 0,
    'Female': 1,
    'Other': 2
})

# exercise frequency
df['Exercise_Frequency'] = df['Exercise_Frequency'].map({
    'Low': 1,
    'Medium': 2,
    'High': 3
})

# diet type
df['Diet_Type'] = df['Diet_Type'].map({
    'Vegetarian': 0,
    'Non-Vegetarian': 1,
    'Vegan': 2
})

# 3.4.3 Generate Boxplot to Check Outlier
numerical_cols = [
    'Age',
    'Sleep_Duration',
    'Stress_Level',
    'Daily_Screen_Time',
    'Caffeine_Intake',
    'Reaction_Time',
    'Memory_Test_Score',
    'Cognitive_Score'
]
for col in numerical_cols:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    save_fig(f"3.4.3_boxplot_{col}")
    plt.show()
    plt.close()

# 3.4.5 Exploratory Data Analysis
# distribution analysis
for col in numerical_cols:

    print("\n" + "="*50)
    print(f"DISTRIBUTION ANALYSIS: {col}")
    print("="*50)

    print(df[col].describe())

    print("\nSkewness:")
    print(df[col].skew())

    print("\nPercentiles:")
    print(df[col].quantile([0.25, 0.5, 0.75]))

    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    save_fig(f"3.4.5_hist_{col}")
    plt.show()
    plt.close()

# =========================================================
# CHAPTER 4.3: OBJECTIVE 1
# =========================================================
print("\n" + "="*60)
print("CHAPTER 4.3: OBJECTIVE 1")
print("="*60)

# 4.3.1 Lifestyle Factors & Sleep Duration
print("\n" + "="*60)
print("LIFESTYLE FACTORS AND SLEEP DURATION")
print("="*60)

sleep_model = smf.ols(
    """
    Sleep_Duration ~
    Exercise_Frequency +
    Stress_Level +
    Daily_Screen_Time +
    Caffeine_Intake +
    Age +
    Gender +
    Diet_Type
    """,
    data=df
).fit()

print(sleep_model.summary())

# 4.3.2 Sleep Duration and Cognitive Performance
print("\n" + "="*60)
print("SLEEP DURATION AND COGNITIVE PERFORMANCE")
print("="*60)

model_with_sleep = smf.ols(
    """
    Cognitive_Score ~
    Sleep_Duration +
    Stress_Level +
    Daily_Screen_Time +
    Exercise_Frequency +
    Caffeine_Intake +
    Age +
    Gender +
    Diet_Type
    """,
    data=df
).fit()

print(model_with_sleep.summary())

# 4.3.3 Lifestyle Factors and Cognitive Performance
# 4.3.3.1 Multiple Linear Regression (without sleep duration)
print("\n" + "="*60)
print("LIFESTYLE FACTORS AND COGNITIVE PERFORMANCE")
print("="*60)

model_no_sleep = smf.ols(
    """
    Cognitive_Score ~
    Stress_Level +
    Daily_Screen_Time +
    Exercise_Frequency +
    Caffeine_Intake +
    Age +
    Gender +
    Diet_Type
    """,
    data=df
).fit()

print(model_no_sleep.summary())

# 4.3.3.2 Effects of Exercise Frequency
# ANOVA
print("\n" + "="*60)
print("ANOVA: EXERCISE FREQUENCY VS COGNITIVE SCORE")
print("="*60)

group_summary = df.groupby(
    'Exercise_Frequency'
)['Cognitive_Score'].agg([
    'mean',
    'std',
    'count'
])
print("\nGroup Summary:")
print(group_summary)

low = df[df['Exercise_Frequency'] == 1]['Cognitive_Score']
medium = df[df['Exercise_Frequency'] == 2]['Cognitive_Score']
high = df[df['Exercise_Frequency'] == 3]['Cognitive_Score']

f_stat, p_value = f_oneway(low, medium, high)

print("F-statistic:", f_stat)
print("P-value:", p_value)

# post-hoc test
print("\n" + "="*60)
print("TUKEY POST-HOC TEST")
print("="*60)

tukey = pairwise_tukeyhsd(
    endog=df['Cognitive_Score'],
    groups=df['Exercise_Frequency'],
    alpha=0.05
)

print(tukey)

#visualise
sns.boxplot(
    x='Exercise_Frequency',
    y='Cognitive_Score',
    data=df
)

plt.title('Cognitive Score by Exercise Frequency')
save_fig(f"4.4.3_anova_exercise")
plt.show()
plt.close()

# 4.3.3.3 Correlation Analysis 
print("\n" + "="*60)
print("CORRELATION ANALYSIS")
print("="*60)

#select variables
correlation_variables = df[[
    'Age',
    'Gender',
    'Sleep_Duration',
    'Stress_Level',
    'Diet_Type',
    'Daily_Screen_Time',
    'Caffeine_Intake',
    'Reaction_Time',
    'Memory_Test_Score',
    'Cognitive_Score'
]]

#calculate correlation matrix
correlation_matrix = correlation_variables.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix.round(3))

#visualise correlation heatmap
plt.figure(figsize=(10,6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title('Correlation Heatmap')
save_fig("4.3.3.3_correlation_analysis")
plt.show()
plt.close()

# 4.3.4 Interaction Effects Btwn Lifestyle Factors
# Lifestyle model: sleep x stress x exercise x screen time
print("\n" + "="*60)
print("SLEEP x STRESS x EXERCISE x SCREEN TIME INTERACTION MODEL")
print("="*60)

interaction_model = smf.ols(
    """
    Cognitive_Score ~
    Sleep_Duration * Stress_Level +
    Sleep_Duration * Exercise_Frequency +
    Sleep_Duration * Daily_Screen_Time
    """,
    data=df
).fit()

print(interaction_model.summary())

# demographic model: sleep x age x gender
print("\n" + "="*60)
print("SLEEP x AGE x GENDER INTERACTION MODEL")
print("="*60)

interaction_demo_model = smf.ols(
    """
    Cognitive_Score ~ 
    Sleep_Duration * Age +
    Sleep_Duration * Gender
    """,
    data=df
).fit()

print(interaction_demo_model.summary())

# 4.3.5 Lifestyle-Based Cluster Analysis
# select numerical features only
cluster_features = df[[
    'Sleep_Duration',
    'Stress_Level',
    'Daily_Screen_Time',
    'Caffeine_Intake',
    'Reaction_Time',
    'Age'
]]

# add categorical variables separately for encoding
cluster_cats = df[['Gender', 'Exercise_Frequency', 'Diet_Type']]

# one-hot encode categorical variables for clustering
cluster_cats_encoded = pd.get_dummies(
    cluster_cats,
    columns=['Gender', 'Exercise_Frequency', 'Diet_Type'],
    drop_first=True
)

# combine numerical & encoded
cluster_features_encoded = pd.concat(
    [cluster_features, cluster_cats_encoded],
    axis=1
)

# standardise variables
scaler = StandardScaler()
scaled_features = scaler.fit_transform(cluster_features_encoded)

print("\nCluster feature shape:", cluster_features_encoded.shape)

# 4.3.5.1 Elbow Method
print("\n" + "="*60)
print("K-MEANS ELBOW METHOD RESULTS")
print("="*60)

inertia = []
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42
    )
    kmeans.fit(scaled_features)
    inertia_value = kmeans.inertia_
    inertia.append(inertia_value)
    
    # print results with actual values
    print(f"k = {k} | Inertia = {inertia_value:.2f}")

# visualise elbow method
plt.figure(figsize=(8,5))
plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.tight_layout()
save_fig("4.3.5.1_elbow")
plt.show()
plt.close()

# 4.3.5.2 Silhouette Scores
print("\n" + "="*60)
print("SILHOUETTE SCORES")
print("="*60)

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42
    )
    labels = kmeans.fit_predict(scaled_features)
    score = silhouette_score(scaled_features, labels)
    silhouette_scores.append(score)

for k, score in zip(range(2, 11), silhouette_scores):
    print(f"k = {k}: {score:.4f}")

# visualisation
plt.figure(figsize=(6,4))
plt.plot(range(2,11), silhouette_scores, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score Method')
save_fig("4.3.5.2_silhouette")
plt.show()
plt.close()

# 4.3.5.3 Final Selected Number of Clusters
print("\n" + "="*60)
print("SELECTION OF NUMBER OF CLUSTERS")
print("="*60)

final_k = 7
print("\nFinal selected number of clusters (based on elbow + silhouette):", final_k)

final_kmeans = KMeans(n_clusters=final_k, random_state=42, n_init=10)
df['Cluster'] = final_kmeans.fit_predict(scaled_features)

final_kmeans_score = silhouette_score(scaled_features, df['Cluster'])
print("\nFinal K-Means Silhouette Score (k=7):", round(final_kmeans_score, 4))

gmm = GaussianMixture(n_components=final_k, random_state=42)
gmm_labels = gmm.fit_predict(scaled_features)
df['GMM_Cluster'] = gmm_labels

gmm_score = silhouette_score(scaled_features, gmm_labels)
print("\nGaussian Mixture Silhouette Score (k=7):", round(gmm_score, 4))

# 4.3.5.4 Cluster Analysis
print("\n" + "="*60)
print("CLUSTER ANALYSIS")
print("="*60)

print("\nCluster Counts:")
print(df['Cluster'].value_counts())

cluster_summary = df.groupby('Cluster')[[
    'Sleep_Duration',
    'Stress_Level',
    'Daily_Screen_Time',
    'Exercise_Frequency',
    'Caffeine_Intake',
    'Cognitive_Score'
]].mean()

print("\nCluster Profiles:")
print(cluster_summary.round(2))

# =========================================================
# CHAPTER 4.4: OBJECTIVE 2
# =========================================================
print("\n" + "="*60)
print("CHAPTER 4.4: OBJECTIVE 2")
print("="*60)

# 4.4.1 Sleep-Cognition Relationship Across Demographic Groups
# demographic stratification (age & gender)
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[18, 25, 35, 45, 60],
    labels=["18-25", "26-35", "36-45", "46-59"]
)

print("\n" + "="*60)
print("SLEEP EFFECT BY AGE + GENDER")
print("="*60)

sleep_demo_effect = df.groupby(["Age_Group", "Gender"])["Cognitive_Score"].mean()
print(sleep_demo_effect)

sleep_corr_demo = df.groupby(["Age_Group", "Gender"]).apply(
    lambda x: x["Sleep_Duration"].corr(x["Cognitive_Score"])
)

print("\nSleep-Cognition Correlation by Demographics:")
print(sleep_corr_demo)

# 4.4.2 Sleep Duration Trend and Distribution of Cognitive Performance
sleep_bins = pd.cut(
    df['Sleep_Duration'],
    bins=[4, 5, 6, 7, 8, 9, 10],
    labels=['4-5', '5-6', '6-7', '7-8', '8-9', '9-10']
)

sleep_bin_summary = df.groupby(
    sleep_bins,
    observed=False
)['Cognitive_Score'].agg([
    'mean',
    'std',
    'count'
])

print("\n" + "="*60)
print("COGNITIVE SCORE BY SLEEP DURATION BINS")
print("="*60)

print(sleep_bin_summary)

#visualisation (linear trend)
sleep_bin_summary['mean'].plot(
    kind='line',
    marker='o'
)

plt.ylabel('Average Cognitive Score')
plt.title('Average Cognitive Score by Sleep Duration Range')
save_fig("4.4.2_cognitive_by_sleep")
plt.show()
plt.close()

# 4.4.3 Modelling the Functional Form of Sleep and Cognitive Performance
# 4.4.3.1 Quadratic Regression
# create squared variable
df['Sleep_Squared'] = df['Sleep_Duration'] ** 2

# quadratic model
quad_model = smf.ols(
    """
    Cognitive_Score ~
    Sleep_Duration + Sleep_Squared
    """,
    data=df
).fit()

print("\n" + "="*60)
print("QUADRATIC MODEL RESULTS")
print("="*60)
print(quad_model.summary())

# visualise quadratic curve
plt.figure(figsize=(6, 4))
sns.regplot(
    x='Sleep_Duration',
    y='Cognitive_Score',
    data=df,
    order=2,
    scatter=False,
    line_kws={'color': 'red', 'linewidth': 2},
    label='Quadratic Trend (95% CI)'
)
  
plt.title('Quadratic Relationship Between Sleep and Cognition')
plt.xlabel('Sleep Duration')
plt.ylabel('Cognitive Score')
plt.legend()
plt.tight_layout()
save_fig("4.4.3.1_quad_reg")
plt.show()
plt.close()

# 4.4.3.2 GAM model
X_gam = df[["Sleep_Duration"]].values
y_gam = df["Cognitive_Score"].values

gam = LinearGAM(s(0)).fit(X_gam, y_gam)

print("\n" + "="*60)
print("GAM MODEL RESULTS")
print("="*60)
print(gam.summary())

# visualisation
XX = np.linspace(df["Sleep_Duration"].min(), df["Sleep_Duration"].max(), 100)
preds = gam.predict(XX)

# calculate the true 95% confidence intervals for the GAM smooth line
conf_intervals = gam.confidence_intervals(XX)

plt.figure(figsize=(6,4))

# plot the GAM trend line
plt.plot(XX, preds, color="red", linewidth=2, label="GAM Fit")

# plot the shaded 95% Confidence Interval band
plt.fill_between(
    XX, 
    conf_intervals[:, 0], 
    conf_intervals[:, 1], 
    color="red", 
    alpha=0.15, 
    label="95% Confidence Interval"
)

plt.title("GAM: Nonlinear Sleep-Cognition Relationship")
plt.xlabel("Sleep Duration")
plt.ylabel("Predicted Cognitive Score")
plt.legend()
plt.tight_layout()
save_fig("4.4.3.2_gam")
plt.show()
plt.close()

# 4.4.4. Estimation of Optimal Sleep Range
# average cognitive score by sleep duration bins
sleep_bins = pd.cut(
    df['Sleep_Duration'],
    bins=[4,5,6,7,8,9,10]
)

sleep_group = df.groupby(
    sleep_bins,
    observed=False
)['Cognitive_Score'].mean()

print("\n" + "="*60)
print("AVERAGE COGNITIVE SCORE BY SLEEP RANGE")
print("="*60)

print(sleep_group)

#visualisation
sleep_group.plot(
    kind='line',
    marker='o'
)
plt.ylabel('Average Cognitive Score')
plt.title('Average Cognitive Score by Sleep Duration Range')
save_fig("4.4.4_avg_cog_vs_sleep")
plt.show()
plt.close()

# =========================================================
# CHAPTER 4.5: OBJECTIVE 3
# =========================================================
print("\n" + "="*60)
print("CHAPTER 4.5: OBJECTIVE 3")
print("="*60)

# 4.5.1 Feature Selection and Data Partitioning
# 4.5.1.1 Feature Selection
X = df[[
    'Sleep_Duration',
    'Stress_Level',
    'Daily_Screen_Time',
    'Exercise_Frequency',
    'Caffeine_Intake',
    'Age',
    'Gender',
    'Diet_Type',
    'Reaction_Time',
    'Memory_Test_Score',
]]

y = df['Cognitive_Score']

# 4.5.1.2 Data Partitioning
split_ratios = [0.1, 0.2, 0.3]
split_results = []

for test_size in split_ratios:

    X_train_tmp, X_test_tmp, y_train_tmp, y_test_tmp = train_test_split(
        X, y,
        test_size=test_size,
        random_state=42
    )

    model = RandomForestRegressor(
        n_estimators=200,
        max_depth=10,
        random_state=42
    )

    model.fit(X_train_tmp, y_train_tmp)
    score = model.score(X_test_tmp, y_test_tmp)

    split_results.append((test_size, score))
    
    print(f"Test size {test_size} → R2: {score:.4f}")

print("\n" + "="*60)
print("DATA SPLIT SENSITIVITY TEST (BASELINE MODEL)")
print("="*60)

BEST_SPLIT = 0.1
print(f"\nBest Split Selected: {BEST_SPLIT}")

# final train validation test split
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y,
    test_size=BEST_SPLIT,
    random_state=42
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp,
    test_size=0.25,
    random_state=42
)

print("\nTraining Set Shape:", X_train.shape)
print("Validation Set Shape:", X_val.shape)
print("Testing Set Shape:", X_test.shape)


# 4.5.2 Baseline Model Comparison
print("\n" + "="*60)
print("MODEL COMPARISON")
print("="*60)

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42)
}

model_scores = []

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_val)
    
    r2 = r2_score(y_val, preds)
    model_scores.append((name, r2))

results_df = pd.DataFrame(model_scores, columns=["Model", "R2"])
best_model_name = results_df.loc[results_df["R2"].idxmax(), "Model"]

# store best baseline model BEFORE tuning
if best_model_name == "Linear Regression":
    selected_model = LinearRegression()

elif best_model_name == "Decision Tree":
    selected_model = DecisionTreeRegressor(random_state=42)

elif best_model_name == "Random Forest":
    selected_model = RandomForestRegressor(random_state=42)
    
# fit selected best model
selected_model.fit(X_train, y_train)

# generate raw predictions on test data
raw_preds = selected_model.predict(X_test)

# bound the predictions strictly between 0-100 because of cognitive performance
final_preds = np.clip(raw_preds, y_test.min(), y_test.max())

# model performance table
final_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42)
}

final_results = []

for name, model in final_models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    preds_clipped = np.clip(preds, y_test.min(), y_test.max())
    
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    r2 = r2_score(y_test, preds)

    final_results.append([name, rmse, r2])

results_df = pd.DataFrame(final_results, columns=["Model", "RMSE", "R2"])

print("\nMODEL COMPARISON TABLE")
print(results_df)

# model performance bar chart
x = np.arange(len(results_df['Model']))
width = 0.35
fig, ax1 = plt.subplots(figsize=(10, 5))

# R² bars
bar1 = ax1.bar(
    x - width/2,
    results_df['R2'],
    width,
    label='R² Score',
    color='steelblue'
)
ax1.set_ylabel('R² Score', color='steelblue')
ax1.tick_params(axis='y', labelcolor='steelblue')
ax1.set_ylim(0, 1)

# RMSE bars
ax2 = ax1.twinx()
bar2 = ax2.bar(
    x + width/2,
    results_df['RMSE'],
    width,
    label='RMSE',
    color='green'
)
ax2.set_ylabel('RMSE', color='green')
ax2.tick_params(axis='y', labelcolor='green')

# X labels
plt.xticks(x, results_df['Model'], rotation=15)
plt.title('Model Performance Comparison (R² vs RMSE)')

# Combined legend
fig.legend([bar1, bar2], ['R² Score', 'RMSE'], loc='center left', bbox_to_anchor=(1.02, 0.5))

plt.tight_layout()
save_fig("4.5.2_model_comparison")
plt.show()
plt.close()

print("\nBest Model Selected:", best_model_name)


# 4.5.4 Model Coefficient Interpretation
print("\n" + "="*60)
print("MODEL COEFFICIENT INTERPRETATION")
print("="*60)

if best_model_name == "Linear Regression":

    coef_df = pd.DataFrame({
        "Feature": X.columns,
        "Coefficient": selected_model.coef_
    }).sort_values(by="Coefficient", ascending=False)

    print("\nLINEAR REGRESSION COEFFICIENTS")
    print(coef_df)

    # visualisatiom
    plt.figure(figsize=(8,5))
    sns.barplot(data=coef_df, x="Coefficient", y="Feature")
    plt.title("Linear Regression Feature Effects on Cognitive Score")
    save_fig("4.5.4_model coefficients")
    plt.show()
    plt.close()

# 4.5.5 Model Performance and Error Analysis
# 4.5.5.1 Residual Analysis
residuals = y_test - final_preds

print("\n" + "="*60)
print("RESIDUAL ANALYSIS")
print("="*60)

print("Mean Residual:", np.mean(residuals))
print("Median Residual:", np.median(residuals))
print("Std Deviation:", np.std(residuals))
print("Min Residual:", np.min(residuals))
print("Max Residual:", np.max(residuals))

print("\n95% of residuals range:")
print(np.percentile(residuals, 2.5), "to", np.percentile(residuals, 97.5))

# visualisation
plt.figure(figsize=(6,4))
plt.scatter(final_preds, residuals, alpha=0.3)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel("Predicted Values")
plt.ylabel("Residuals")
plt.title("Residual Plot (Predicted vs Error)")
save_fig("4.5.5.1_residuals")
plt.show()
plt.close()

# 4.5.5.2 Error Distribution Summary
print("\n" + "="*60)
print("ERROR DISTRIBUTION SUMMARY")
print("="*60)

print("Skewness:", residuals.skew())
print("Kurtosis:", residuals.kurtosis())

print("\nQuantiles:")
print("25%:", residuals.quantile(0.25))
print("50% (Median):", residuals.quantile(0.50))
print("75%:", residuals.quantile(0.75))

print("\n95% Error Range:")
print(residuals.quantile(0.025), "to", residuals.quantile(0.975))

print("\nMean Absolute Error (manual check):", np.mean(np.abs(residuals)))

# visualisation
plt.figure(figsize=(6,4))
sns.histplot(residuals, bins=30, kde=True)
plt.title("Distribution of Prediction Errors (Residuals)")
plt.xlabel("Error")
plt.ylabel("Frequency")
save_fig("4.5.5.2_error_dist")
plt.show()
plt.close()


# 4.5.6 Feature Importance and Predictive Drivers
importance_df = None 

print("\nNOTE: Linear Regression selected as best model.\nIt does not provide built-in feature importance.")

sorted_models = results_df.sort_values(by="R2", ascending=False)
second_best_name = sorted_models.iloc[1]["Model"]

print("\nSecond Best Model Selected (Proxy for Interpretability):", second_best_name)

if second_best_name == "Decision Tree":
    proxy_model = DecisionTreeRegressor(random_state=42)

elif second_best_name == "Random Forest":
    proxy_model = RandomForestRegressor(random_state=42)

else:
    proxy_model = RandomForestRegressor(random_state=42)

proxy_model.fit(X_train, y_train)

importances = proxy_model.feature_importances_

importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importances
}).sort_values(by="Importance", ascending=False)

print("\n" + "="*60)
print("FEATURE IMPORTANCE")
print("="*60)

print(importance_df)

# visualisation
sns.barplot(data=importance_df, x="Importance", y="Feature")
plt.title("Feature Importance (Proxy Model)")
save_fig("4.5.6_feature_importance")
plt.show()
plt.close()

# 4.5.7 Prediction Summary and Final Model Behaviour
corr = np.corrcoef(y_test, final_preds)[0, 1]

print("\n=" + "="*60)
print("PREDICTION CORRELATION")
print("="*60)
print("Correlation (Actual vs Predicted):", corr)

# visualisation
plt.figure(figsize=(6,6))
plt.scatter(y_test, final_preds, alpha=0.3)
plt.xlabel("Actual Cognitive Score")
plt.ylabel("Predicted Cognitive Score")
plt.title("Final Model: Actual vs Predicted")
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         'r--')
save_fig("4.5.7_pred_correlation")
plt.show()
plt.close()


print("ALL DONE")
```
